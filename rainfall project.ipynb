```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import os
os.chdir(r"C:\Users\Hinduja\OneDrive\Desktop\python")
```

```python
rainfall = pd.read_csv('rainfall.csv')
rainfall
```

```python

plt.figure(figsize=[8,6])
chennai_rain = rainfall['Chennai']

avg_rainfall = chennai_rain.mean()
threshold = avg_rainfall * 1.25




plt.axhline(
    y=threshold,
    color='red',
    linestyle='--',
    linewidth=2,
    label='High Rainfall Threshold (Chennai)'
)
high_risk = chennai_rain > threshold

plt.plot(
    rainfall.Year,
    chennai_rain,
    marker='o',
    label='Yearly Rainfall (Chennai)'
)
plt.xticks(rainfall.Year[::3])

plt.scatter(
    rainfall.Year[high_risk],
    chennai_rain[high_risk],
    color='red',
    s=90,
    zorder=5,
    label='High-Risk Years (Chennai)')

plt.title('High Rainfall Anomaly Detection for Chennai', fontsize=20,color='green',fontdict={'fontstyle':'italic'})
plt.xlabel('Rainfall year')
plt.ylabel('Annual rainfall(cm)')

plt.grid(True,linestyle='--',alpha=0.5)
plt.gca().set_axisbelow(True)
plt.legend()
plt.savefig('Chennai_rainfall_intensity.png',dpi=300)
plt.show()


```
